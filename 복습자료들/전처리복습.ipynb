• ## 문제 1. 영어 문장 토큰화

  아래 텍스트를 NLTK의 `sent_tokenize()`로 문장 단위로 분리하세요.

  ```python
  review_text = (
      "Dr. Lee studies NLP at SK AI. "
      "He likes tokenization! "
      "Does he use Python?"
  )

  조건:

  - 분리된 문장 목록을 sentences 변수에 저장하세요.
  - 각 문장을 번호와 함께 출력하세요.
  - 전체 문장 수도 출력하세요.

  확인할 점:

  - Dr. 뒤에서 문장이 잘못 분리되지 않는지 확인하세요.

  힌트:

  sentences = sent_tokenize(...)


````python
review_text = (
    "Dr. Lee studies NLP at SK AI. "
    "He likes tokenization! "
    "Does he use Python?"
)

sentences = sent_tokenize(review_text)
print(sentences)
print("총 길이 : ", len(sentences))
for index, sentence in enumerate(sentences, start=1):
    print(f"{index}. {sentence}")


````

````python
import kss

mixed_text = (
    "오늘 NLP study를 시작했다. "
    "I will train a tokenizer tomorrow! "
    '"이 모델 supports Korean and English?"라고 민지가 물었다. '
    "답은 Yes, it does."
)

sentences = kss.split_sentences(mixed_text)

for index, sentence in enumerate(sentences, start=1):
    print(f"{index}. {sentence}")
````

## 문제 2. 단어 토큰과 문자 토큰 비교

  문제 1에서 사용한 원본 문자열 `review_text`를 서로 다른 단위로 나누어 비교하세요.

  조건:

  - `review_text`를 `word_tokenize()`에 전달하고, 결과를 `word_tokens` 변수에 저장하세요.
  - 같은 원본 문자열 `review_text`를 `list()`에 전달하고, 결과를 `character_tokens` 변수에 저장하세요.
  - 두 결과의 토큰 목록과 토큰 수를 각각 출력하세요.
  - 문자 토큰 결과에 공백 문자 `' '`가 포함되는지 확인하고 주석으로 작성하세요.

  힌트:

  - `word_tokenize()`는 단어·구두점 단위로 나눕니다.
  - 문자열에 `list()`를 적용하면 문자 한 글자씩 나뉩니다.

````python


word_tokens = word_tokenize(review_text)
character_tokens = list(review_text)
print(word_tokens)
print(character_tokens)

for token in character_tokens:
    if token == " ":
        print("공백 있음")
        break

print("단어 토큰 수:", len(word_tokens))
print("문자 토큰 수:", len(character_tokens))
````
